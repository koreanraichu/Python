# 불러온 데이터
- 이것도 그려보시지! 

```python
load_df = pl.read_csv('data/Aspirin.csv', separator=';', null_values="")
load_df
```

## Matplotlib

```python
plt.scatter(load_df['HBA'], load_df['HBD'])
plt.show()
```

- 왜 됨? 

## Seaborn

```python
sns.scatterplot(load_df.to_pandas(), x = 'HBA', y = 'HBD')
plt.show()
```

- Seaborn은 한번에 안됐는데 둘이 뭔 차이인지는 모르겠고요... 일단 HBA, HBD에 널은 두개씩 있네요. (널때문에 한번에 안돼서 판다스로 변환함)

```python
load_df.null_count()
```

## Plotly

```python
fig = px.scatter(load_df, x = load_df['HBA'], y = load_df['HBD'])
fig.show()
```

- 저 0은... 널이여? 

```python
trouble_maker = load_df.filter((pl.col("HBA") == 0) & (pl.col("HBD") == 0))
trouble_maker
```
