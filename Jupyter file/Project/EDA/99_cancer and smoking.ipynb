# GEOparse로 암종별 데이터 받아서 분석해보기 
## 개요
우리 전공을 하다 보면 제일 많이 다루는게 암, 알츠하이머, 파킨슨입니다. 그 중에서도 암 하면서 제일 많이 나오는 변이원이 담배인데... 아주 담배는 진짜 심심하면 나와요... 교수님들도 담배 피지 말라고 하시는데... 그럼에도 피는 분이 계세요. 수업중에 헉하면서도... 그죠. 니코틴이 무섭습니다. 

정말 담배는 백해무익할까? 에 대한 결론을 내리기 위한 분석입니다. 굳이 흡연자들만 많이 걸리는 암이 아니라 메이저한 암들(얘: 위암, 대장암)을 묶어서... 
GEOparse로 함 털어봅시다. 그리고 진짜로 흡연이 백해무익한지 데이터로 확인해봅시다. 

## 프로젝트 정보
- 인원: 1인(개인 프로젝트)
- 버전: 3.10(TF_base)
- 설치할 것들: Biopython, GEOparse(NCBI GEO에 접근할 때 필요)
- 데이터 리소스: GEOparse

        - 비흡연자: #FDD835
