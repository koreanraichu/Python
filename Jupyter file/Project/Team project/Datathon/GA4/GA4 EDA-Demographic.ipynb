# 데이터 열기 & 전처리
- not set: 측정기준과 관련하여 애널리틱스로 들어온 정보가 없을 때 사용되는 자리표시자 이름 (이라고 구글이 그랬어요)
- 여기서는 결측값으로 간주하고 언노운으로 처리하겠습니다. (아직 논의된 바 없음...)

```python
def split_ga4_csv(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    tables = []
    current_table = []

    for line in lines:
        # 줄바꿈만 있거나 쉼표만 있는 빈 줄을 만나면 지금까지의 데이터를 저장
        if line.strip() == "" or line.strip().replace(',', '') == "":
            if current_table:
                # 메모리 상에서 해당 부분만 데이터프레임으로 변환
                df = pd.read_csv(io.StringIO("".join(current_table)), comment='#')
                tables.append(df)
                current_table = []
        else:
            current_table.append(line)

    # 마지막 테이블 처리
    if current_table:
        tables.append(pd.read_csv(io.StringIO("".join(current_table))))

    return tables

# 사용 예시
all_tables = split_ga4_csv('data/20260319_Rextreme_GA4/Generate_leads_overview.csv')

# GA4 overview 파일의 경우 보통:
# all_tables[0] -> 상단 요약 정보 (2컬럼)
# all_tables[1] -> 중간 상세 내역 (N컬럼)
```

## 칼럼들의 의미
1. Country: 국가
2. Active users: 활성 유저
3. New users: 신규 유저
4. Engaged sessions: 의미 있는 세션 수
5. Engagement rate: 참여율
6. Engaged sessions per active user: 유저 1명당 평균 참여 세션 수
7. Average engagement time per active user: 유저당 평균 체류시간
8. Event count: 총 이벤트 발생 수
9. Key events: 중요한 이벤트
10. User key event rate: 유저 기준 전환율 (몇명이나 목표 행동을 했는가?)

```python
demographic_df
```

```python
clean_df = clean_df.groupby('isKorea').agg({
    'Active users': 'sum',           # 2번: 총 유입
    'New users': 'sum',              # 3번: 신규 유입
    'Engaged sessions': 'sum',       # 4번: 유의미한 방문 합계
    'Engagement rate': 'mean',       # 5번: 평균 참여율
    'Engaged sessions per active user':'mean', # 6번: 활성 유저당 평균 참여율
    'Average engagement time per active user': 'mean', # 7번: 평균 체류시간
    'Event count': 'sum',             # 8번: 총 활동량
    'Key events':'sum'                # 9번: 키 이벤트
}).reset_index()
clean_df = clean_df.sort_values('isKorea', ascending=False)
```
