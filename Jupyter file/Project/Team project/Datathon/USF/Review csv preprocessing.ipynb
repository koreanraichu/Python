# 날짜 형식 변환
- YYYY년 MM월 DD일->YYYY-MM-DD

```python
df_name = [fork_0, fork_1, fork_2, fork_3]

for i in range(len(df_name)):
    # 1. 날짜 변환 (이건 그대로 두셔도 됩니다)
    df_name[i]['날짜'] = pd.to_datetime(df_name[i]['날짜'])

    # 2. 정렬 후 반드시 reset_index(drop=True)를 실행해서 덮어씌우기
    # .sort_values() 뒤에 바로 붙여서 리스트 요소를 교체하는 게 가장 깔끔합니다.
    df_name[i] = df_name[i].sort_values('날짜').reset_index(drop=True)

# 3. 개별 변수 업데이트 (fork_0 ~ fork_3 이름표 갱신)
fork_0, fork_1, fork_2, fork_3 = df_name
```

```python
fork_1
```

# 합! 체!

```python
df_concat = pd.concat([fork_0, fork_1], axis = 0)
df_concat = pd.concat([df_concat, fork_2], axis = 0)
df_concat = pd.concat([df_concat, fork_3], axis = 0)
```
