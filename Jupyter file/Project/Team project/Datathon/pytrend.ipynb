```python
from pytrends.request import TrendReq
import pandas as pd

# 구글 연결
pytrends = TrendReq(hl='ko-KR', tz=540)

# 키워드 설정 (한 번에 5개까지 가능)
kw_list = ["렉스트림", "REXTREME", "세방전지", "로케트배터리"]

# 페이로드 구축 (기간: 2025-12-24 ~ 2026-02-24)
pytrends.build_payload(kw_list, cat=0, timeframe='2025-12-24 2026-02-24', geo='KR')

# 시간대별 관심도 가져오기
data = pytrends.interest_over_time()
print(data.head())
```
