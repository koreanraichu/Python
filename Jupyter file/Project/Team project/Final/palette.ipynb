```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# 커스텀 팔레트
custom_cmap = sns.blend_palette(['#002D72', '#f1f0ec','#ffb659'], as_cmap=True)

# 컬러맵 프리뷰용
data = np.random.randn(10, 10)
sns.heatmap(data, cmap=custom_cmap)
plt.show()
```

```python
# 커스텀 팔레트
custom_cmap = sns.blend_palette(['#002D72', '#040303','#ffb659'], as_cmap=True)

# 컬러맵 프리뷰용
data = np.random.randn(10, 10)
sns.heatmap(data, cmap=custom_cmap)
plt.show()
```

```python
custom_cmap = sns.diverging_palette(360, 265, center="dark", as_cmap=True)

# 컬러맵 프리뷰용
data = np.random.randn(10, 10)
sns.heatmap(data, cmap=custom_cmap)
plt.show()
```
