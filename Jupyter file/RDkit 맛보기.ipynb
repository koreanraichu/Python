```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd
import plotly.express as px
```

```python
smiles_list = ["CC(=O)Nc1ccc(O)cc1", "CC(C)Cc1ccc(cc1)[C@@H](C)C(=O)O", "COc1cc2ccc(cc2cc1)[C@@H](C)C(=O)O", "O=C(C)Oc1ccccc1C(=O)O"]

data = []
for smi in smiles_list:
    mol = Chem.MolFromSmiles(smi)
    data.append({
        "smiles": smi,
        "MW": Descriptors.MolWt(mol),
        "LogP": Descriptors.MolLogP(mol)
    })
```

```python
df = pd.DataFrame(data)

fig = px.scatter(df, x="MW", y="LogP", text="smiles")
fig.show()
```
