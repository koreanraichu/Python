```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import chisquare
from scipy.stats import chi2_contingency
```

# Chi-square
- 카이제곱 검정이라고도 하는데 저는 걍 카이스퀘어라고 합니다. Chi는 그리스 알파벳 $\Chi$인데 이걸 자판으로 어떻게 씁니까... 뭐 본인이 그리스인이라면 모를까. 그래서 Chi로 쓰는겁니다. 
- 카이제곱 검정은 범주형 데이터(명목 척도)의 관찰 빈도가 기대 빈도와 유의미한 차이가 있는지, 또는 두 변수가 독립적인지 분석하는 통계적 가설 검정입니다. 아니 이게 뭔 소리임? 
    - 예를 들어보겠습니다. 성별과 특정 주류에 대한 선호도에 관련이 있을까요, 없을까요? 아니면 성별과 흡연률은? 

## 카이제곱 검정의 종류
1. 동질성 검정: 공대랑 자연대랑 성비가 비슷할까요? ~~사실 과바과임~~
2. 독립성 검정: 얘(범주형 1)랑 쟤(범주형 2)랑 관련이 있을까요? 
3. 적합도 검정: 관측된 어떤 분포가 특정 분포와 유의미한 차이가 있을까요? 

# 적합성 검정
- 이게 이 분포와 유의미한 차이가 있나? 
- 귀무가설: 관측된 분포가 특정 분포와 유의미한 차이가 없다.
- 대립가설: 관측된 분포가 특정 분포와 유의미한 차이가 있다.

```python
# 관찰 빈도 (실제 데이터)
observed = [12, 8, 11, 9, 10, 10] 
# 기대 빈도 (이론적 데이터: 60번 던졌으니 각각 10번)
expected = [10, 10, 10, 10, 10, 10]

# 검정
chi_stat, p_val = chisquare(f_obs=observed, f_exp=expected)

if p_val < 0.05:
    print("귀무가설 기각")
else:
    print("귀무가설 수립")
```

- 아, 해봤더니 이론적인거(1/6씩 이븐하게 나오는)랑 유의미한 차이가 없구나... 

```python
# 관찰 빈도 (실제 데이터)
observed = [12, 8, 11, 19, 10, 0] 
# 기대 빈도 (이론적 데이터: 60번 던졌으니 각각 10번)
expected = [10, 10, 10, 10, 10, 10]

# 검정
chi_stat, p_val = chisquare(f_obs=observed, f_exp=expected)

if p_val < 0.05:
    print("귀무가설 기각")
else:
    print("귀무가설 수립")
```

- 이게 만약 6이 안 나오게 조작된 주사위라면 이렇게 나오는겁니다. 
- 가챠게임류 뻑하면 확률 주작 논란 터지잖아요? 그럴 때 이거 써먹으셔도 됩니다. 

# 독립성 검정
- 귀무가설: 두 범주형 변수가 서로 독립적이다.
- 대립가설: 두 범주형 변수가 서로 독립적인 관계가 아니다. 

```python
# 교차표(Contingency Table) 생성
# 행: 성별, 열: 선호 브랜드
data = pd.DataFrame({
    'Brand_A': [20, 30], # 남성 중 A선호, 여성 중 A선호
    'Brand_B': [15, 10]  # 남성 중 B선호, 여성 중 B선호
}, index=['Male', 'Female'])

# 검정 (correction=False는 연속성 보정 미적용)
chi_stat, p_val, dof, expected = chi2_contingency(data)

if p_val < 0.05:
    print("귀무가설 기각: 이 두 변수가 독립적이지 않다.")
else:
    print("귀무가설 수립: 이 두 변수가 독립적이다.")
```

- 귀무가설이 기각됐다면 '아 얘랑 쟤랑 관련이 있구나' 하시면 됩니다. 
- 위 예시의 경우 '특정 브랜드의 선호도가 성별과 관련이 있다'는 의미입니다. 

# 동질성 검정
- 귀무가설: 두 집단의 분포가 동일하다. 
- 대립가설: 두 집단의 분포가 동일하지 않다. 

```python
# 계산 방식은 독립성 검정과 동일하게 chi2_contingency를 사용합니다.
# 데이터: 서울(흡연 40, 비흡연 60) vs 부산(흡연 30, 비흡연 70)
city_data = [[40, 60], [30, 70]]
chi_stat, p_val, _, _ = chi2_contingency(city_data)

if p_val < 0.05:
    print("귀무가설 기각: 두 집단의 분포가 동일하지 않다. ")
else:
    print("귀무가설 수립: 두 집단의 분포가 동일하다. ")
```

- 서울과 부산의 흡연/비흡연자 분포가 비슷한 걸로 나왔네요. 

```python
# 계산 방식은 독립성 검정과 동일하게 chi2_contingency를 사용합니다.
# 집단 1과 집단 2의 성비 
gender_ratio = [[93, 7], [70, 30]]
chi_stat, p_val, _, _ = chi2_contingency(gender_ratio)

if p_val < 0.05:
    print("귀무가설 기각: 두 집단의 분포가 동일하지 않다. ")
else:
    print("귀무가설 수립: 두 집단의 분포가 동일하다. ")
```

- 가상의 데이터이긴 하지만 두 집단의 성비가 '다른' 것으로 나옵니다.
